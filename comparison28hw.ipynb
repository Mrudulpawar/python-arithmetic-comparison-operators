## Comparison Operators
#### >, <, >=, <=, ==, !=...

## using data types 
#### Int, Float, Complex, Bool, String...



## (A) Using Int

### 1...Int-Int

```python
a=10
b=5
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

### 2...Int-Float

```python
a=10
b=5.5
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

### 3...Int-Complex

```python
a=10
b=5+5j
###print(a>b)  type error
###print(a<b)
###print(a>=b)
###print(a<=b)
print(a==b)
print(a!=b)
```

### 4...Int-Bool

```python
a=10
b=True
print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
```

### 5...Int-String

```python
a=10
b="Python"
###print(a>b)  type error
###print(a<b)
###print(a>=b)
###print(a<=b)
print(a==b)
print(a!=b)
```
