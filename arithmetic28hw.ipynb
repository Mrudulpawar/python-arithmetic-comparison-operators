## Arithmetic Operators
#### +, -, *, /, %, //...

## using data types 
#### Int, Float, Complex, Bool, String...



## (A) Using Int

### 1...Int-Int

```python
a=10
b=5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
```

### 2...Int-Float

```python
a=10
b=2.5
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
```

### 3...Int-Complex

```python
a=10
b=5+2j
print(a+b)
print(a-b)
print(a*b)
print(a/b)
```

### 4...Int-Bool

```python
a=10
b=True
print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b)
print(a//b)
```

### 5...Int-String

```python
a = 5
b = "Python"
print(str(a) +b)
print(a*b)
```

## (B) Using Float

### 3...Float-Bool

### 4...Float-String

```python
a=10.5
b="Python"
print(str(a) +b)
```

## (D) Using bool
